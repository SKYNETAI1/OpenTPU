# Chat with Qwen3.5-2B on Ultra96

This notebook uses the public Python API, the separately downloaded GGUF model, the bundled tokenizer, and the prebuilt Ultra96 overlay. It does not download files at run time. Run the cells in order.

```python
from pathlib import Path
import sys

PACKAGE_DIR = Path("/home/xilinx/jupyter_notebooks/qwen35_public")
MODEL_PATH = PACKAGE_DIR / "Qwen3.5-2B-Q3_K_S.gguf"
TOKENIZER_PATH = PACKAGE_DIR

for required in (MODEL_PATH, TOKENIZER_PATH / "tokenizer.json"):
    if not required.is_file():
        raise FileNotFoundError(required)

if str(PACKAGE_DIR) not in sys.path:
    sys.path.insert(0, str(PACKAGE_DIR))

print(f"Model: {MODEL_PATH} ({MODEL_PATH.stat().st_size / 1024**2:.1f} MiB)")
print(f"Tokenizer: {TOKENIZER_PATH / 'tokenizer.json'}")
```

```python
from qwen35 import ChatSession

old_session = globals().get("session")
if old_session is not None:
    old_session.close()

session = ChatSession(
    model_path=MODEL_PATH,
    tokenizer_path=TOKENIZER_PATH,
    max_context=256,
)
print("Model ready.")
```

```python
result = session.ask(
    "What is 2 to the power of 3?",
    max_new_tokens=32,
)
result
```

Run the previous cell again with a new prompt to continue the conversation. Use `session.clear()` to clear history and `session.close()` before restarting the notebook kernel.
